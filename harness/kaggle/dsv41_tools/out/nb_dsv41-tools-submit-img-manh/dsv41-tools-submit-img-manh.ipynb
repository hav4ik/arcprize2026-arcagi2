```python
# Cell 2r (E066 image candidate): two runtime patches into Manh's installed vLLM (wheels g5e1bccd07c .. g71dca8bc24), embedded
# (internet is off) and sha256-guarded:
# - vllm/models/deepseek_v41/nvidia/model.py: his Engram row-chunk fix a6c3ee8c2b (the Kaggle Models store Engram as
#   row chunks) plus per-layer routed-expert counts (`n_routed_experts_per_layer`, branch dynamic-count-moe b3a3f642a9);
# - vllm/model_executor/layers/fused_moe/router/dsv4_topk.py: the Triton router's expert-count whitelist widened from
#   (128, 256, 384) to 6..512 (the CUDA fallback rejects per-layer counts such as 72 or 152).
# Per file: target hash already -> nothing to do; base hash (5e1bccd07c) -> apply, check the target hash, drop the
# file's __pycache__ entry; any other hash -> commit runs RAISE; the competition rerun leaves the file as is (a
# per-layer model then fails to load and the fallback submission stands). Records go to {WORK}/e066_vllm_patch.json (cell 4b prints them; any action
# other than patched / already_fixed fails the commit run).
import base64, glob, hashlib, importlib.metadata as _md, re

IMG_PATCHES = [  # per file: the known wheels' originals -> our target (row chunks are in his wheels from f506afa210)
    {"rel": "vllm/models/deepseek_v41/nvidia/model.py", "variants": [
        {"wheel": "g5e1bccd07c", "base": "ead1e087c29b7d9ef76313e9849cffecfb01bec50d8625d4724527af853d10dd",
         "fixed": "75f3725adc5d131ece87a0334774683e41a027e77c37ecec4ff62a61f1ad7f42",
         "diff_sha": "6241ce4ee6c42068437d219a49d4feb980f49e632676f346af974df436582561", "diff": base64.b64decode("ZGlmZiAtLWdpdCBhL3ZsbG0vbW9kZWxzL2RlZXBzZWVrX3Y0MS9udmlkaWEvbW9kZWwucHkgYi92bGxtL21vZGVscy9kZWVwc2Vla192NDEvbnZpZGlhL21vZGVsLnB5CmluZGV4IDgxZTg0MzMzYzQuLjEwMjQ5Yzk0NTEgMTAwNjQ0Ci0tLSBhL3ZsbG0vbW9kZWxzL2RlZXBzZWVrX3Y0MS9udmlkaWEvbW9kZWwucHkKKysrIGIvdmxsbS9tb2RlbHMvZGVlcHNlZWtfdjQxL252aWRpYS9tb2RlbC5weQpAQCAtMTQzLDYgKzE0MywzOCBAQCBpZiB0eXBpbmcuVFlQRV9DSEVDS0lORzoKIGxvZ2dlciA9IGluaXRfbG9nZ2VyKF9fbmFtZV9fKQogCiAKK2RlZiByb3V0ZWRfZXhwZXJ0c19wZXJfbGF5ZXIoY29uZmlnKSAtPiBsaXN0W2ludF0gfCBOb25lOgorICAgICIiIlBlci1sYXllciByb3V0ZWQtZXhwZXJ0IGNvdW50cyBvZiBhIHBydW5lZCBleHBvcnQsIG9yIE5vbmUgd2hlbiBldmVyeQorICAgIGRlY29kZXIgbGF5ZXIgaGFzIGBgbl9yb3V0ZWRfZXhwZXJ0c2BgLgorCisgICAgRXhwb3J0cyB0aGF0IGtlZXAgYSBkaWZmZXJlbnQgbnVtYmVyIG9mIGV4cGVydHMgaW4gZWFjaCBNb0UgbGF5ZXIgbGlzdAorICAgIHRoZW0gaW4gYGBuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllcmBgIChvbmUgZW50cnkgcGVyIGRlY29kZXIgbGF5ZXIsIGluCisgICAgbGF5ZXIgb3JkZXIpOyBlYWNoIGxheWVyJ3MgZ2F0ZSwgY29ycmVjdGlvbiBiaWFzLCB2aXNpb24gYmlhcyBhbmQgZXhwZXJ0CisgICAgdGVuc29ycyBhcmUgc3RvcmVkIGF0IHRoYXQgbGF5ZXIncyBjb3VudC4KKyAgICAiIiIKKyAgICBjb3VudHMgPSBnZXRhdHRyKGNvbmZpZywgIm5fcm91dGVkX2V4cGVydHNfcGVyX2xheWVyIiwgTm9uZSkKKyAgICBpZiBjb3VudHMgaXMgTm9uZToKKyAgICAgICAgcmV0dXJuIE5vbmUKKyAgICBjb3VudHMgPSBbaW50KGMpIGZvciBjIGluIGNvdW50c10KKyAgICBpZiBsZW4oY291bnRzKSAhPSBjb25maWcubnVtX2hpZGRlbl9sYXllcnM6CisgICAgICAgIHJhaXNlIFZhbHVlRXJyb3IoCisgICAgICAgICAgICBmIm5fcm91dGVkX2V4cGVydHNfcGVyX2xheWVyIGhhcyB7bGVuKGNvdW50cyl9IGVudHJpZXMsIGV4cGVjdGVkICIKKyAgICAgICAgICAgIGYibnVtX2hpZGRlbl9sYXllcnM9e2NvbmZpZy5udW1faGlkZGVuX2xheWVyc30iCisgICAgICAgICkKKyAgICBpZiBtaW4oY291bnRzKSA8IGNvbmZpZy5udW1fZXhwZXJ0c19wZXJfdG9rOgorICAgICAgICByYWlzZSBWYWx1ZUVycm9yKAorICAgICAgICAgICAgZiJuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllciBoYXMgYSBsYXllciB3aXRoIHttaW4oY291bnRzKX0gZXhwZXJ0cywgIgorICAgICAgICAgICAgZiJmZXdlciB0aGFuIG51bV9leHBlcnRzX3Blcl90b2s9e2NvbmZpZy5udW1fZXhwZXJ0c19wZXJfdG9rfSIKKyAgICAgICAgKQorICAgIHJldHVybiBjb3VudHMKKworCitkZWYgbWF4X3JvdXRlZF9leHBlcnRzKGNvbmZpZykgLT4gaW50OgorICAgICIiIlRoZSBsYXJnZXN0IGRlY29kZXItbGF5ZXIgcm91dGVkLWV4cGVydCBjb3VudCAoZXhwZXJ0LW5hbWUgbWFwcGluZykuIiIiCisgICAgY291bnRzID0gcm91dGVkX2V4cGVydHNfcGVyX2xheWVyKGNvbmZpZykKKyAgICByZXR1cm4gY29uZmlnLm5fcm91dGVkX2V4cGVydHMgaWYgY291bnRzIGlzIE5vbmUgZWxzZSBtYXgoY291bnRzKQorCisKIGNsYXNzIERlZXBzZWVrVjRNb0UoRGVlcHNlZWtWNE1vRUJhc2UpOgogICAgIGRlZiBfX2luaXRfXygKICAgICAgICAgc2VsZiwKQEAgLTE1Miw5ICsxODQsMjIgQEAgY2xhc3MgRGVlcHNlZWtWNE1vRShEZWVwc2Vla1Y0TW9FQmFzZSk6CiAgICAgICAgIHJlZHVjZV9yZXN1bHRzOiBib29sID0gVHJ1ZSwKICAgICApOgogICAgICAgICBjb25maWcgPSB2bGxtX2NvbmZpZy5tb2RlbF9jb25maWcuaGZfY29uZmlnCisgICAgICAgIGxheWVyX2lkeCA9IGV4dHJhY3RfbGF5ZXJfaW5kZXgocHJlZml4KQorICAgICAgICBwZXJfbGF5ZXIgPSByb3V0ZWRfZXhwZXJ0c19wZXJfbGF5ZXIoY29uZmlnKQogICAgICAgICBuX3JvdXRlZF9leHBlcnRzID0gY29uZmlnLm5fcm91dGVkX2V4cGVydHMKKyAgICAgICAgaWYgcGVyX2xheWVyIGlzIG5vdCBOb25lIGFuZCBsYXllcl9pZHggPCBjb25maWcubnVtX2hpZGRlbl9sYXllcnM6CisgICAgICAgICAgICBuX3JvdXRlZF9leHBlcnRzID0gcGVyX2xheWVyW2xheWVyX2lkeF0KKyAgICAgICAgaWYgKAorICAgICAgICAgICAgcGVyX2xheWVyIGlzIG5vdCBOb25lCisgICAgICAgICAgICBhbmQgbGVuKHNldChwZXJfbGF5ZXIpKSA+IDEKKyAgICAgICAgICAgIGFuZCB2bGxtX2NvbmZpZy5wYXJhbGxlbF9jb25maWcuZW5hYmxlX2VwbGIKKyAgICAgICAgKToKKyAgICAgICAgICAgIHJhaXNlIE5vdEltcGxlbWVudGVkRXJyb3IoCisgICAgICAgICAgICAgICAgIkVQTEIgbmVlZHMgdGhlIHNhbWUgcm91dGVkLWV4cGVydCBjb3VudCBpbiBldmVyeSBNb0UgbGF5ZXI7ICIKKyAgICAgICAgICAgICAgICAidGhpcyBleHBvcnQgc2V0cyBuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllci4iCisgICAgICAgICAgICApCiAgICAgICAgIG5fYWN0aXZhdGVkX2V4cGVydHMgPSBjb25maWcubnVtX2V4cGVydHNfcGVyX3RvawotICAgICAgICBpZiBleHRyYWN0X2xheWVyX2luZGV4KHByZWZpeCkgPj0gY29uZmlnLm51bV9oaWRkZW5fbGF5ZXJzOgorICAgICAgICBpZiBsYXllcl9pZHggPj0gY29uZmlnLm51bV9oaWRkZW5fbGF5ZXJzOgogICAgICAgICAgICAgbl9yb3V0ZWRfZXhwZXJ0cyA9ICgKICAgICAgICAgICAgICAgICBnZXRhdHRyKGNvbmZpZywgImRzcGFya19uX3JvdXRlZF9leHBlcnRzIiwgMCkgb3Igbl9yb3V0ZWRfZXhwZXJ0cwogICAgICAgICAgICAgKQpAQCAtMTU2OCw3ICsxNjEzLDkgQEAgY2xhc3MgRGVlcHNlZWtWNE1vZGVsKG5uLk1vZHVsZSwgRWFnbGVNb2RlbE1peGluKToKICAgICBkZWYgZ2V0X2V4cGVydF9tYXBwaW5nKHNlbGYpIC0+IGxpc3RbdHVwbGVbc3RyLCBzdHIsIGludCwgc3RyXV06CiAgICAgICAgIGZpcnN0X2xheWVyID0gbmV4dChpdGVyKGlzbGljZShzZWxmLmxheWVycywgc2VsZi5zdGFydF9sYXllciwgc2VsZi5lbmRfbGF5ZXIpKSkKICAgICAgICAgaWYgZmlyc3RfbGF5ZXIuZmZuLnVzZV9uYXRpdmVfbWVnYV9tb2U6Ci0gICAgICAgICAgICByZXR1cm4gbWFrZV9kZWVwc2Vla192NF9leHBlcnRfcGFyYW1zX21hcHBpbmcoc2VsZi5jb25maWcubl9yb3V0ZWRfZXhwZXJ0cykKKyAgICAgICAgICAgIHJldHVybiBtYWtlX2RlZXBzZWVrX3Y0X2V4cGVydF9wYXJhbXNfbWFwcGluZygKKyAgICAgICAgICAgICAgICBtYXhfcm91dGVkX2V4cGVydHMoc2VsZi5jb25maWcpCisgICAgICAgICAgICApCiAgICAgICAgICMgUGFyYW1zIGZvciB3ZWlnaHRzLCBmcDggd2VpZ2h0IHNjYWxlcywgZnA4IGFjdGl2YXRpb24gc2NhbGVzCiAgICAgICAgICMgKHBhcmFtX25hbWUsIHdlaWdodF9uYW1lLCBleHBlcnRfaWQsIHNoYXJkX2lkKQogICAgICAgICByZXR1cm4gZnVzZWRfbW9lX21ha2VfZXhwZXJ0X3BhcmFtc19tYXBwaW5nKApAQCAtMTU3Niw3ICsxNjIzLDcgQEAgY2xhc3MgRGVlcHNlZWtWNE1vZGVsKG5uLk1vZHVsZSwgRWFnbGVNb2RlbE1peGluKToKICAgICAgICAgICAgIGNrcHRfZ2F0ZV9wcm9qX25hbWU9IncxIiwKICAgICAgICAgICAgIGNrcHRfZG93bl9wcm9qX25hbWU9IncyIiwKICAgICAgICAgICAgIGNrcHRfdXBfcHJval9uYW1lPSJ3MyIsCi0gICAgICAgICAgICBudW1fZXhwZXJ0cz1zZWxmLmNvbmZpZy5uX3JvdXRlZF9leHBlcnRzLAorICAgICAgICAgICAgbnVtX2V4cGVydHM9bWF4X3JvdXRlZF9leHBlcnRzKHNlbGYuY29uZmlnKSwKICAgICAgICAgICAgIG51bV9yZWR1bmRhbnRfZXhwZXJ0cz1maXJzdF9sYXllci5mZm4ubl9yZWR1bmRhbnRfZXhwZXJ0cywKICAgICAgICAgKQogCkBAIC0xNjEwLDYgKzE2NTcsNjcgQEAgZGVmIF9saW5lYXJfc2NhbGVfcGFyYW1fbmFtZSh2bGxtX2NvbmZpZzogVmxsbUNvbmZpZywgZXhwZXJ0X2R0eXBlOiBzdHIpIC0+IHN0cjoKICAgICByZXR1cm4gIndlaWdodF9zY2FsZSIgaWYgdXNlX214ZnA4IGVsc2UgIndlaWdodF9zY2FsZV9pbnYiCiAKIAorX0VOR1JBTV9ST1dfQ0hVTksgPSByZS5jb21waWxlKHIiXiguKlwuZW5ncmFtKVwuZW1iZWRcLih3ZWlnaHR8c2NhbGUpXC5yb3dzXyhcZCspJCIpCisKKworZGVmIF9lbmdyYW1fcm93X2NodW5rcygKKyAgICB3ZWlnaHRzOiBJdGVyYWJsZVt0dXBsZVtzdHIsIHRvcmNoLlRlbnNvcl1dLAorICAgIG1hcHBlcjogV2VpZ2h0c01hcHBlciwKKyAgICBwYXJhbXM6IGRpY3Rbc3RyLCBubi5QYXJhbWV0ZXJdLAorICAgIGxvYWRlZDogc2V0W3N0cl0sCispIC0+IEl0ZXJhYmxlW3R1cGxlW3N0ciwgdG9yY2guVGVuc29yXV06CisgICAgIiIiTG9hZCBFbmdyYW0gdGFibGVzIHN0b3JlZCBhcyByb3cgY2h1bmtzLgorCisgICAgRXhwb3J0cyB3aXRoIGEgZmlsZSBzaXplIGxpbWl0IHN0b3JlIGVhY2ggdGFibGUgYXMKKyAgICBgYDxsYXllcj4uZW5ncmFtLmVtYmVkLnt3ZWlnaHQsc2NhbGV9LnJvd3NfPHN0YXJ0PmBgLiBFYWNoIGNodW5rJ3Mgb3ZlcmxhcAorICAgIHdpdGggdGhpcyByYW5rJ3Mgcm93cyBpcyBjb3BpZWQgaW50byB0aGUgdGFibGU7IG90aGVyIHdlaWdodHMgcGFzcworICAgIHRocm91Z2guIEEgdGFibGUgdGhhdCByZWNlaXZlcyBhbnkgY2h1bmsgbXVzdCBoYXZlIHRoaXMgcmFuaydzIHJvd3MKKyAgICBjb3ZlcmVkIGV4YWN0bHkgb25jZS4KKyAgICAiIiIKKyAgICBzcGFuczogZGljdFtzdHIsIGxpc3RbdHVwbGVbaW50LCBpbnRdXV0gPSB7fQorICAgIGZvciBuYW1lLCB0ZW5zb3IgaW4gd2VpZ2h0czoKKyAgICAgICAgbWF0Y2ggPSBfRU5HUkFNX1JPV19DSFVOSy5tYXRjaChuYW1lKQorICAgICAgICBpZiBtYXRjaCBpcyBOb25lOgorICAgICAgICAgICAgeWllbGQgbmFtZSwgdGVuc29yCisgICAgICAgICAgICBjb250aW51ZQorICAgICAgICAjIE1hcCBvbmx5IHRoZSBtb2R1bGUgcHJlZml4OiB0aGUgVkwgd3JhcHBlciBoYW5kcyBpdHMgY2hpbGQgbmFtZXMgdGhhdAorICAgICAgICAjIGFyZSBhbHJlYWR5IG1hcHBlZCwgYW5kIHRoZSBzdWZmaXggcnVsZXMgZG8gbm90IHNlZSBgYC5yb3dzXzxzdGFydD5gYC4KKyAgICAgICAgbW9kdWxlID0gbWFwcGVyLm1hcF9uYW1lKG1hdGNoLmdyb3VwKDEpKQorICAgICAgICBpZiBtb2R1bGUgaXMgTm9uZToKKyAgICAgICAgICAgIGNvbnRpbnVlCisgICAgICAgIGxlYWYgPSAid2VpZ2h0IiBpZiBtYXRjaC5ncm91cCgyKSA9PSAid2VpZ2h0IiBlbHNlICJ3ZWlnaHRfc2NhbGVfaW52IgorICAgICAgICBwbmFtZSA9IGYie21vZHVsZX0uZW1iZWRfdG9rZW5zLntsZWFmfSIKKyAgICAgICAgcGFyYW0gPSBwYXJhbXMuZ2V0KHBuYW1lKQorICAgICAgICBpZiBwYXJhbSBpcyBOb25lOgorICAgICAgICAgICAgaWYgYW55KGtleS5zdGFydHN3aXRoKGYie21vZHVsZX0uIikgZm9yIGtleSBpbiBwYXJhbXMpOgorICAgICAgICAgICAgICAgIHJhaXNlIFZhbHVlRXJyb3IoZiJFbmdyYW0gcm93IGNodW5rIHtuYW1lfSBtYXRjaGVzIG5vIHBhcmFtZXRlciIpCisgICAgICAgICAgICBjb250aW51ZSAgIyBBIGxheWVyIG9mIGFub3RoZXIgcGlwZWxpbmUtcGFyYWxsZWwgcmFuay4KKyAgICAgICAgaWYgdGVuc29yLmR0eXBlID09IHRvcmNoLmZsb2F0OF9lOG0wZm51OgorICAgICAgICAgICAgdGVuc29yID0gdGVuc29yLnZpZXcodG9yY2gudWludDgpCisgICAgICAgIHN0YXJ0LCBsbyA9IGludChtYXRjaC5ncm91cCgzKSksIHBhcmFtLmVuZ3JhbV92b2NhYl9zdGFydAorICAgICAgICBhID0gbWF4KHN0YXJ0LCBsbykKKyAgICAgICAgYiA9IG1pbihzdGFydCArIHRlbnNvci5zaGFwZVswXSwgbG8gKyBwYXJhbS5zaGFwZVswXSkKKyAgICAgICAgcm93cyA9IHNwYW5zLnNldGRlZmF1bHQocG5hbWUsIFtdKQorICAgICAgICBpZiBhIDwgYjoKKyAgICAgICAgICAgIHBhcmFtLmRhdGFbYSAtIGxvIDogYiAtIGxvXS5jb3B5Xyh0ZW5zb3JbYSAtIHN0YXJ0IDogYiAtIHN0YXJ0XSkKKyAgICAgICAgICAgIHJvd3MuYXBwZW5kKChhIC0gbG8sIGIgLSBsbykpCisgICAgICAgIGxvYWRlZC5hZGQocG5hbWUpCisgICAgZm9yIHBuYW1lLCByb3dzIGluIHNwYW5zLml0ZW1zKCk6CisgICAgICAgIGVuZCA9IDAKKyAgICAgICAgZm9yIGEsIGIgaW4gc29ydGVkKHJvd3MpOgorICAgICAgICAgICAgaWYgYSAhPSBlbmQ6CisgICAgICAgICAgICAgICAgcmFpc2UgVmFsdWVFcnJvcigKKyAgICAgICAgICAgICAgICAgICAgZiJFbmdyYW0gcm93IGNodW5rcyBvZiB7cG5hbWV9OiByYW5rIHJvd3MgIgorICAgICAgICAgICAgICAgICAgICBmInttaW4oYSwgZW5kKX0uLnttYXgoYSwgZW5kKX0gbWlzc2luZyBvciBkdXBsaWNhdGVkIgorICAgICAgICAgICAgICAgICkKKyAgICAgICAgICAgIGVuZCA9IGIKKyAgICAgICAgaWYgZW5kICE9IHBhcmFtc1twbmFtZV0uc2hhcGVbMF06CisgICAgICAgICAgICByYWlzZSBWYWx1ZUVycm9yKAorICAgICAgICAgICAgICAgIGYiRW5ncmFtIHJvdyBjaHVua3MgY292ZXIge2VuZH0gb2YgIgorICAgICAgICAgICAgICAgIGYie3BhcmFtc1twbmFtZV0uc2hhcGVbMF19IHJhbmsgcm93cyBvZiB7cG5hbWV9IgorICAgICAgICAgICAgKQorCisKIGRlZiBfbWFrZV9kZWVwc2Vla192NF93ZWlnaHRzX21hcHBlcigKICAgICBleHBlcnRfZHR5cGU6IHN0ciwgbGluZWFyX3NjYWxlX25hbWU6IHN0ciA9ICJ3ZWlnaHRfc2NhbGVfaW52IgogKSAtPiBXZWlnaHRzTWFwcGVyOgpAQCAtMTg1OCw4ICsxOTY2LDE2IEBAIGNsYXNzIERlZXBzZWVrVjQxTExNRm9yQ2F1c2FsTE0oCiAgICAgICAgIGlmIGdwdHFfZmlsZXM6CiAgICAgICAgICAgICAjIEFscmVhZHkgaW4gdkxMTSBuYW1pbmcsIHdoaWNoIGJvdGggdGhlIHRleHQgYW5kIFZMIG1hcHBlcnMga2VlcC4KICAgICAgICAgICAgIHdlaWdodHMgPSBjaGFpbih3ZWlnaHRzLCBpdGVyX2V4cGVydF9ncHRxX3dlaWdodHMoZ3B0cV9maWxlcywgIm1vZGVsLiIpKQorICAgICAgICBlbmdyYW1fY2h1bmtzOiBzZXRbc3RyXSA9IHNldCgpCisgICAgICAgIHdlaWdodHMgPSBfZW5ncmFtX3Jvd19jaHVua3MoCisgICAgICAgICAgICB3ZWlnaHRzLAorICAgICAgICAgICAgc2VsZi5oZl90b192bGxtX21hcHBlciwKKyAgICAgICAgICAgIGRpY3Qoc2VsZi5uYW1lZF9wYXJhbWV0ZXJzKCkpLAorICAgICAgICAgICAgZW5ncmFtX2NodW5rcywKKyAgICAgICAgKQogICAgICAgICBsb2FkZXIgPSBBdXRvV2VpZ2h0c0xvYWRlcihzZWxmKQogICAgICAgICBsb2FkZWRfcGFyYW1zID0gbG9hZGVyLmxvYWRfd2VpZ2h0cyh3ZWlnaHRzLCBtYXBwZXI9c2VsZi5oZl90b192bGxtX21hcHBlcikKKyAgICAgICAgbG9hZGVkX3BhcmFtcyB8PSBlbmdyYW1fY2h1bmtzCiAgICAgICAgIHNlbGYucHJvY2Vzc193ZWlnaHRzX2FmdGVyX2xvYWRpbmcoKQogICAgICAgICBpZiBjb25maWcuZW5ncmFtX2NvbmZpZyBhbmQgY29uZmlnLmVuZ3JhbV9jb25maWcudXNlX3RocDoKICAgICAgICAgICAgICMgTG9hZGluZyB3ZWlnaHRzIHJlZmlsbHMgdGhlIGZpbGUgY2FjaGU7IHJlbGVhc2UgaXQgYmVmb3JlIE1BRFZfQ09MTEFQU0UuCg==").decode()},
        {"wheel": "gf506afa210 / g71dca8bc24", "base": "029aa62e383332aad37836866f4aebe8859a277977ee368b44f6f1dd8cc8ad64",
         "fixed": "ad4ac0fd9bc604feb43afc7cc6052cc749b914653d731020e8f28074a078bd37",
         "diff_sha": "a164071c3f446a159c5e69e0a334fca69cb27112007371f1efb199b31cf058c2", "diff": base64.b64decode("ZGlmZiAtLWdpdCBhL3ZsbG0vbW9kZWxzL2RlZXBzZWVrX3Y0MS9udmlkaWEvbW9kZWwucHkgYi92bGxtL21vZGVscy9kZWVwc2Vla192NDEvbnZpZGlhL21vZGVsLnB5CmluZGV4IGU1M2FiZWI3ODUuLjc2MmNlNzYzOGUgMTAwNjQ0Ci0tLSBhL3ZsbG0vbW9kZWxzL2RlZXBzZWVrX3Y0MS9udmlkaWEvbW9kZWwucHkKKysrIGIvdmxsbS9tb2RlbHMvZGVlcHNlZWtfdjQxL252aWRpYS9tb2RlbC5weQpAQCAtMTU0LDYgKzE1NCwzOCBAQCBpZiB0eXBpbmcuVFlQRV9DSEVDS0lORzoKIGxvZ2dlciA9IGluaXRfbG9nZ2VyKF9fbmFtZV9fKQogCiAKK2RlZiByb3V0ZWRfZXhwZXJ0c19wZXJfbGF5ZXIoY29uZmlnKSAtPiBsaXN0W2ludF0gfCBOb25lOgorICAgICIiIlBlci1sYXllciByb3V0ZWQtZXhwZXJ0IGNvdW50cyBvZiBhIHBydW5lZCBleHBvcnQsIG9yIE5vbmUgd2hlbiBldmVyeQorICAgIGRlY29kZXIgbGF5ZXIgaGFzIGBgbl9yb3V0ZWRfZXhwZXJ0c2BgLgorCisgICAgRXhwb3J0cyB0aGF0IGtlZXAgYSBkaWZmZXJlbnQgbnVtYmVyIG9mIGV4cGVydHMgaW4gZWFjaCBNb0UgbGF5ZXIgbGlzdAorICAgIHRoZW0gaW4gYGBuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllcmBgIChvbmUgZW50cnkgcGVyIGRlY29kZXIgbGF5ZXIsIGluCisgICAgbGF5ZXIgb3JkZXIpOyBlYWNoIGxheWVyJ3MgZ2F0ZSwgY29ycmVjdGlvbiBiaWFzLCB2aXNpb24gYmlhcyBhbmQgZXhwZXJ0CisgICAgdGVuc29ycyBhcmUgc3RvcmVkIGF0IHRoYXQgbGF5ZXIncyBjb3VudC4KKyAgICAiIiIKKyAgICBjb3VudHMgPSBnZXRhdHRyKGNvbmZpZywgIm5fcm91dGVkX2V4cGVydHNfcGVyX2xheWVyIiwgTm9uZSkKKyAgICBpZiBjb3VudHMgaXMgTm9uZToKKyAgICAgICAgcmV0dXJuIE5vbmUKKyAgICBjb3VudHMgPSBbaW50KGMpIGZvciBjIGluIGNvdW50c10KKyAgICBpZiBsZW4oY291bnRzKSAhPSBjb25maWcubnVtX2hpZGRlbl9sYXllcnM6CisgICAgICAgIHJhaXNlIFZhbHVlRXJyb3IoCisgICAgICAgICAgICBmIm5fcm91dGVkX2V4cGVydHNfcGVyX2xheWVyIGhhcyB7bGVuKGNvdW50cyl9IGVudHJpZXMsIGV4cGVjdGVkICIKKyAgICAgICAgICAgIGYibnVtX2hpZGRlbl9sYXllcnM9e2NvbmZpZy5udW1faGlkZGVuX2xheWVyc30iCisgICAgICAgICkKKyAgICBpZiBtaW4oY291bnRzKSA8IGNvbmZpZy5udW1fZXhwZXJ0c19wZXJfdG9rOgorICAgICAgICByYWlzZSBWYWx1ZUVycm9yKAorICAgICAgICAgICAgZiJuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllciBoYXMgYSBsYXllciB3aXRoIHttaW4oY291bnRzKX0gZXhwZXJ0cywgIgorICAgICAgICAgICAgZiJmZXdlciB0aGFuIG51bV9leHBlcnRzX3Blcl90b2s9e2NvbmZpZy5udW1fZXhwZXJ0c19wZXJfdG9rfSIKKyAgICAgICAgKQorICAgIHJldHVybiBjb3VudHMKKworCitkZWYgbWF4X3JvdXRlZF9leHBlcnRzKGNvbmZpZykgLT4gaW50OgorICAgICIiIlRoZSBsYXJnZXN0IGRlY29kZXItbGF5ZXIgcm91dGVkLWV4cGVydCBjb3VudCAoZXhwZXJ0LW5hbWUgbWFwcGluZykuIiIiCisgICAgY291bnRzID0gcm91dGVkX2V4cGVydHNfcGVyX2xheWVyKGNvbmZpZykKKyAgICByZXR1cm4gY29uZmlnLm5fcm91dGVkX2V4cGVydHMgaWYgY291bnRzIGlzIE5vbmUgZWxzZSBtYXgoY291bnRzKQorCisKIGNsYXNzIERlZXBzZWVrVjRNb0UoRGVlcHNlZWtWNE1vRUJhc2UpOgogICAgIGRlZiBfX2luaXRfXygKICAgICAgICAgc2VsZiwKQEAgLTE2Myw5ICsxOTUsMjIgQEAgY2xhc3MgRGVlcHNlZWtWNE1vRShEZWVwc2Vla1Y0TW9FQmFzZSk6CiAgICAgICAgIHJlZHVjZV9yZXN1bHRzOiBib29sID0gVHJ1ZSwKICAgICApOgogICAgICAgICBjb25maWcgPSB2bGxtX2NvbmZpZy5tb2RlbF9jb25maWcuaGZfY29uZmlnCisgICAgICAgIGxheWVyX2lkeCA9IGV4dHJhY3RfbGF5ZXJfaW5kZXgocHJlZml4KQorICAgICAgICBwZXJfbGF5ZXIgPSByb3V0ZWRfZXhwZXJ0c19wZXJfbGF5ZXIoY29uZmlnKQogICAgICAgICBuX3JvdXRlZF9leHBlcnRzID0gY29uZmlnLm5fcm91dGVkX2V4cGVydHMKKyAgICAgICAgaWYgcGVyX2xheWVyIGlzIG5vdCBOb25lIGFuZCBsYXllcl9pZHggPCBjb25maWcubnVtX2hpZGRlbl9sYXllcnM6CisgICAgICAgICAgICBuX3JvdXRlZF9leHBlcnRzID0gcGVyX2xheWVyW2xheWVyX2lkeF0KKyAgICAgICAgaWYgKAorICAgICAgICAgICAgcGVyX2xheWVyIGlzIG5vdCBOb25lCisgICAgICAgICAgICBhbmQgbGVuKHNldChwZXJfbGF5ZXIpKSA+IDEKKyAgICAgICAgICAgIGFuZCB2bGxtX2NvbmZpZy5wYXJhbGxlbF9jb25maWcuZW5hYmxlX2VwbGIKKyAgICAgICAgKToKKyAgICAgICAgICAgIHJhaXNlIE5vdEltcGxlbWVudGVkRXJyb3IoCisgICAgICAgICAgICAgICAgIkVQTEIgbmVlZHMgdGhlIHNhbWUgcm91dGVkLWV4cGVydCBjb3VudCBpbiBldmVyeSBNb0UgbGF5ZXI7ICIKKyAgICAgICAgICAgICAgICAidGhpcyBleHBvcnQgc2V0cyBuX3JvdXRlZF9leHBlcnRzX3Blcl9sYXllci4iCisgICAgICAgICAgICApCiAgICAgICAgIG5fYWN0aXZhdGVkX2V4cGVydHMgPSBjb25maWcubnVtX2V4cGVydHNfcGVyX3RvawotICAgICAgICBpZiBleHRyYWN0X2xheWVyX2luZGV4KHByZWZpeCkgPj0gY29uZmlnLm51bV9oaWRkZW5fbGF5ZXJzOgorICAgICAgICBpZiBsYXllcl9pZHggPj0gY29uZmlnLm51bV9oaWRkZW5fbGF5ZXJzOgogICAgICAgICAgICAgbl9yb3V0ZWRfZXhwZXJ0cyA9ICgKICAgICAgICAgICAgICAgICBnZXRhdHRyKGNvbmZpZywgImRzcGFya19uX3JvdXRlZF9leHBlcnRzIiwgMCkgb3Igbl9yb3V0ZWRfZXhwZXJ0cwogICAgICAgICAgICAgKQpAQCAtMTgxNyw3ICsxODYyLDkgQEAgY2xhc3MgRGVlcHNlZWtWNE1vZGVsKG5uLk1vZHVsZSwgRWFnbGVNb2RlbE1peGluKToKICAgICBkZWYgZ2V0X2V4cGVydF9tYXBwaW5nKHNlbGYpIC0+IGxpc3RbdHVwbGVbc3RyLCBzdHIsIGludCwgc3RyXV06CiAgICAgICAgIGZpcnN0X2xheWVyID0gbmV4dChpdGVyKGlzbGljZShzZWxmLmxheWVycywgc2VsZi5zdGFydF9sYXllciwgc2VsZi5lbmRfbGF5ZXIpKSkKICAgICAgICAgaWYgZmlyc3RfbGF5ZXIuZmZuLnVzZV9uYXRpdmVfbWVnYV9tb2U6Ci0gICAgICAgICAgICByZXR1cm4gbWFrZV9kZWVwc2Vla192NF9leHBlcnRfcGFyYW1zX21hcHBpbmcoc2VsZi5jb25maWcubl9yb3V0ZWRfZXhwZXJ0cykKKyAgICAgICAgICAgIHJldHVybiBtYWtlX2RlZXBzZWVrX3Y0X2V4cGVydF9wYXJhbXNfbWFwcGluZygKKyAgICAgICAgICAgICAgICBtYXhfcm91dGVkX2V4cGVydHMoc2VsZi5jb25maWcpCisgICAgICAgICAgICApCiAgICAgICAgICMgUGFyYW1zIGZvciB3ZWlnaHRzLCBmcDggd2VpZ2h0IHNjYWxlcywgZnA4IGFjdGl2YXRpb24gc2NhbGVzCiAgICAgICAgICMgKHBhcmFtX25hbWUsIHdlaWdodF9uYW1lLCBleHBlcnRfaWQsIHNoYXJkX2lkKQogICAgICAgICByZXR1cm4gZnVzZWRfbW9lX21ha2VfZXhwZXJ0X3BhcmFtc19tYXBwaW5nKApAQCAtMTgyNSw3ICsxODcyLDcgQEAgY2xhc3MgRGVlcHNlZWtWNE1vZGVsKG5uLk1vZHVsZSwgRWFnbGVNb2RlbE1peGluKToKICAgICAgICAgICAgIGNrcHRfZ2F0ZV9wcm9qX25hbWU9IncxIiwKICAgICAgICAgICAgIGNrcHRfZG93bl9wcm9qX25hbWU9IncyIiwKICAgICAgICAgICAgIGNrcHRfdXBfcHJval9uYW1lPSJ3MyIsCi0gICAgICAgICAgICBudW1fZXhwZXJ0cz1zZWxmLmNvbmZpZy5uX3JvdXRlZF9leHBlcnRzLAorICAgICAgICAgICAgbnVtX2V4cGVydHM9bWF4X3JvdXRlZF9leHBlcnRzKHNlbGYuY29uZmlnKSwKICAgICAgICAgICAgIG51bV9yZWR1bmRhbnRfZXhwZXJ0cz1maXJzdF9sYXllci5mZm4ubl9yZWR1bmRhbnRfZXhwZXJ0cywKICAgICAgICAgKQogCg==").decode()}]},
    {"rel": "vllm/model_executor/layers/fused_moe/router/dsv4_topk.py", "variants": [
        {"wheel": "g5e1bccd07c .. g71dca8bc24", "base": "7e1e101ab2fb266bef34ac5826749a9803add4cde84eeb68093882948a14c50d",
         "fixed": "ab32ea7e2f3a69767d996cefb3840cca18c0bbc926a98616534be04c32c6d005",
         "diff_sha": "f8dda98c795b443f1d7f9117f76f0176287f8b1e63d82ac11fb620a04a4c7e6c", "diff": base64.b64decode("ZGlmZiAtLWdpdCBhL3ZsbG0vbW9kZWxfZXhlY3V0b3IvbGF5ZXJzL2Z1c2VkX21vZS9yb3V0ZXIvZHN2NF90b3BrLnB5IGIvdmxsbS9tb2RlbF9leGVjdXRvci9sYXllcnMvZnVzZWRfbW9lL3JvdXRlci9kc3Y0X3RvcGsucHkKaW5kZXggOWMxZWJhOTZlMi4uYmVmNDMzOWY3OSAxMDA2NDQKLS0tIGEvdmxsbS9tb2RlbF9leGVjdXRvci9sYXllcnMvZnVzZWRfbW9lL3JvdXRlci9kc3Y0X3RvcGsucHkKKysrIGIvdmxsbS9tb2RlbF9leGVjdXRvci9sYXllcnMvZnVzZWRfbW9lL3JvdXRlci9kc3Y0X3RvcGsucHkKQEAgLTEyLDYgKzEyLDE1IEBAIGZyb20gdmxsbS5wbGF0Zm9ybXMgaW1wb3J0IGN1cnJlbnRfcGxhdGZvcm0KIGZyb20gdmxsbS50cml0b25fdXRpbHMgaW1wb3J0IHRsLCB0cml0b24KIAogX1RPUEsgPSA2CisjIFRoZSBrZXJuZWwgaG9sZHMgYSBnYXRpbmcgcm93IGluIG9uZSB3YXJwJ3MgcmVnaXN0ZXJzIChCTE9DS19OID0gbmV4dCBwb3dlciBvZgorIyB0d28sIGxhbmVzID49IG51bV9leHBlcnRzIG1hc2tlZCksIHNvIGFueSBjb3VudCBmcm9tIHRvcC1rIHRvIDUxMiB3b3Jrcy4KKyMgUHJ1bmVkIGV4cG9ydHMgd2l0aCBwZXItbGF5ZXIgZXhwZXJ0IGNvdW50cyAoZS5nLiA3Mi0yMjQgcm91dGVkIGV4cGVydHMpIHRha2UKKyMgdGhpcyBwYXRoIHRvbzsgdGhlIENVREEgdG9wa19zb2Z0cGx1c19zcXJ0IGZhbGxiYWNrIG9ubHkgYWNjZXB0cyBhIGZpeGVkIGxpc3QuCitfTUFYX0VYUEVSVFMgPSA1MTIKKworCitkZWYgX3N1cHBvcnRzX251bV9leHBlcnRzKG51bV9leHBlcnRzOiBpbnQpIC0+IGJvb2w6CisgICAgcmV0dXJuIF9UT1BLIDw9IG51bV9leHBlcnRzIDw9IF9NQVhfRVhQRVJUUwogCiAjIEFkYXB0ZWQgZnJvbToKICMgaHR0cHM6Ly9naXRodWIuY29tL3NnbC1wcm9qZWN0L3NnbGFuZy9ibG9iL21haW4vcHl0aG9uL3NnbGFuZy9qaXRfa2VybmVsL21vZV9mdXNlZF9nYXRlLnB5CkBAIC0yOCw3ICszNyw3IEBAIGRlZiBjYW5fdXNlX2RzdjRfdG9waygKICAgICAgICAgY3VycmVudF9wbGF0Zm9ybS5pc19jdWRhKCkKICAgICAgICAgYW5kIGdhdGluZ19vdXRwdXQuZHR5cGUgPT0gdG9yY2guZmxvYXQzMgogICAgICAgICBhbmQgZ2F0aW5nX291dHB1dC5uZGltID09IDIKLSAgICAgICAgYW5kIGdhdGluZ19vdXRwdXQuc2hhcGVbMV0gaW4gKDEyOCwgMjU2LCAzODQpCisgICAgICAgIGFuZCBfc3VwcG9ydHNfbnVtX2V4cGVydHMoZ2F0aW5nX291dHB1dC5zaGFwZVsxXSkKICAgICAgICAgYW5kIGdhdGluZ19vdXRwdXQuaXNfY29udGlndW91cygpCiAgICAgICAgIGFuZCBjb3JyZWN0aW9uX2JpYXMgaXMgbm90IE5vbmUKICAgICAgICAgYW5kIGNvcnJlY3Rpb25fYmlhcy5kdHlwZSA9PSB0b3JjaC5mbG9hdDMyCkBAIC0xODYsNyArMTk1LDcgQEAgZGVmIHJlZ2lzdGVyX2RzdjRfdG9wa193YXJtdXAoCiAgICAgIiIiCiAgICAgaWYgKAogICAgICAgICBjdXJyZW50X3BsYXRmb3JtLmlzX2N1ZGEoKQotICAgICAgICBhbmQgbnVtX2V4cGVydHMgaW4gKDEyOCwgMjU2LCAzODQpCisgICAgICAgIGFuZCBfc3VwcG9ydHNfbnVtX2V4cGVydHMobnVtX2V4cGVydHMpCiAgICAgICAgIGFuZCB0b3BrID09IF9UT1BLCiAgICAgICAgIGFuZCByZW5vcm1hbGl6ZQogICAgICk6Cg==").decode()}]},
]


def apply_unified(text, diff):
    """Apply a one-file unified diff to `text`; raises ValueError when a context or removed line does not match."""
    src, out, i = text.splitlines(keepends=True), [], 0
    lines = diff.splitlines(keepends=True)
    k = 0
    while k < len(lines):
        m = re.match(r"^@@ -(\d+)(?:,(\d+))? \+\d+(?:,\d+)? @@", lines[k])
        k += 1
        if not m:
            continue
        start = int(m.group(1)) - 1 if m.group(2) != "0" else int(m.group(1))
        if start < i:
            raise ValueError("overlapping hunks")
        out += src[i:start]
        i = start
        while k < len(lines) and not lines[k].startswith("@@"):
            tag, body = lines[k][:1], lines[k][1:]
            if tag in (" ", "-"):
                if i >= len(src) or src[i] != body:
                    raise ValueError(f"hunk mismatch at line {i + 1}: {body[:60]!r}")
                if tag == " ":
                    out.append(body)
                i += 1
            elif tag == "+":
                out.append(body)
            elif tag != "\\":
                break
            k += 1
    return "".join(out + src[i:])


def guarded_patch(path, rerun, variants):
    """-> record {file, before, after, action, wheel}; the variant whose base hash matches is applied; a file already at
    a target hash is left alone; raises in commit runs on an unknown file or a failed patch."""
    rec = {"file": str(path), "bases": [v["base"][:16] for v in variants]}
    try:
        data = open(path, "rb").read()
        rec["before"] = hashlib.sha256(data).hexdigest()
        if rec["before"] in {v["fixed"] for v in variants}:
            rec["action"] = "already_fixed"
            return rec
        v = next((v for v in variants if v["base"] == rec["before"]), None)
        if v is None:
            raise ValueError(f"unknown file (sha256 {rec['before'][:16]}), not a known wheel's original")
        rec["wheel"] = v["wheel"]
        new = apply_unified(data.decode(), v["diff"]).encode()
        if hashlib.sha256(new).hexdigest() != v["fixed"]:
            raise ValueError("patched file does not have the target hash")
        tmp = f"{path}.img_tmp"
        with open(tmp, "wb") as f:
            f.write(new)
        os.replace(tmp, path)
        rec["after"] = hashlib.sha256(open(path, "rb").read()).hexdigest()
        stem = os.path.splitext(os.path.basename(path))[0]
        rec["pyc_removed"] = []
        for pyc in glob.glob(os.path.join(os.path.dirname(path), "__pycache__", f"{stem}.*.pyc")):
            os.remove(pyc)
            rec["pyc_removed"].append(os.path.basename(pyc))
        rec["action"] = "patched"
        return rec
    except Exception as exc:
        rec["action"] = "failed"
        rec["error"] = f"{type(exc).__name__}: {exc}"
        if not rerun:
            raise RuntimeError(f"E066 VLLM PATCH FAILED (commit run): {rec}") from exc
        rec["action"] = "rerun_left_as_is"
        return rec


IMG_PATCH_RECORDS = []
ROWCHUNK = {"action": "skipped (mode fallback)"}
if MODE != "fallback":
    for _p in IMG_PATCHES:
        for _v in _p["variants"]:
            assert hashlib.sha256(_v["diff"].encode()).hexdigest() == _v["diff_sha"], f"embedded diff corrupted: {_p['rel']}"
        try:
            _fp = str(_md.distribution("vllm").locate_file(_p["rel"]))
        except Exception as exc:
            _fp = f"<vllm not installed: {type(exc).__name__}>"
        _rec = guarded_patch(_fp, RERUN, _p["variants"])
        IMG_PATCH_RECORDS.append(_rec)
        print(("E066 VLLM PATCH " + _rec["action"].upper()).replace("_", " "), _rec, flush=True)
        if _rec["action"] == "rerun_left_as_is":
            print("=" * 100, f"E066 VLLM PATCH FALLBACK (rerun): {_p['rel']} left as is; a per-layer model may not "
                  "load and the fallback submission would stand", "=" * 100, sep="\n", flush=True)
    ROWCHUNK = {**IMG_PATCH_RECORDS[0], "all": IMG_PATCH_RECORDS}
    try:
        import json as _pj, os as _po
        with open(_po.path.join(WORK, "e066_vllm_patch.json"), "w") as _pf:
            _pj.dump(IMG_PATCH_RECORDS, _pf, indent=1)
    except Exception as exc:
        print("E066 VLLM PATCH: record not written:", type(exc).__name__, exc, flush=True)
```

```python
# Cell 2s (E066, 2026-10-29): restart supervisor for Manh's server (cell 2m). Without it, one engine death ends the
# run's new attempts (B32-M's 46.5 fits a stop around Kaggle hour 4-6). A daemon thread watches MANH_SERVER_PID; when it
# exits, or makes no progress for E066_STALL_S while requests are running, the thread:
#   1. writes the status "restarting" (the solver's S26 then stops launching, and re-queues in-flight attempts);
#   2. stops the server's process group (TERM, then KILL) and waits until it and every GPU compute process are gone, so
#      two engines' host RAM never overlap;
#   3. reruns cell 2m's exact launcher command (_cmd, _env) with a fresh --output-root, dsv41_logs/r<N>;
#   4. waits for /v1/models to list dsv41, then writes the status "up" with the new start's pid, log and startup time.
# The status file uses the format of Manh's --supervise (PR #2), which the pinned solver's S26 follows
# (ARC_EXTERNAL_SERVER_STATUS, cell 3). It gives up, with the status "failed", after E066_MAX_RESTARTS restarts or when
# less than E066_RESTART_MIN_LEFT_S remain before stop-new. Cell 3s stops it before cell 4 stops the server. Uses cell
# 0's T0, WORK, MODE and cell 2m's _cmd, _env, MANH_BASE_URL, MANH_SERVER_PID, MANH_SERVER_LOG, _manh_alive, _t.
import json
import os
import signal
import subprocess
import threading
import time
import urllib.request

SUPERVISOR_STATUS = f"{WORK}/dsv41_logs/e066_supervisor.json"
SUPERVISOR = {"stop": False, "restarts": 0, "events": []}
E066_MAX_RESTARTS = int(os.environ.get("E066_MAX_RESTARTS", "2"))
E066_RESTART_MIN_LEFT_S = float(os.environ.get("E066_RESTART_MIN_LEFT_S", "3600"))
E066_STALL_S = float(os.environ.get("E066_STALL_S", "1500"))
E066_UP_TIMEOUT_S = float(os.environ.get("E066_UP_TIMEOUT_S", "3000"))


def _sup_log(msg):
    line = f"[+{time.time() - T0:.0f}s] E066 SUPERVISOR: {msg}"
    SUPERVISOR["events"].append(line)
    print(line, flush=True)


def _sup_write(**status):
    tmp = SUPERVISOR_STATUS + ".tmp"
    with open(tmp, "w") as f:
        json.dump({**status, "t": round(time.time() - T0)}, f)
    os.replace(tmp, SUPERVISOR_STATUS)


def _sup_group_alive(pgid):
    """Any non-zombie process left in process group pgid."""
    for d in os.listdir("/proc"):
        if d.isdigit():
            try:
                with open(f"/proc/{d}/stat") as f:
                    rest = f.read().rsplit(")", 1)[1].split()
                if int(rest[2]) == pgid and rest[0] not in ("Z", "X"):
                    return True
            except (OSError, ValueError, IndexError):
                pass
    return False


def _sup_gpu_apps():
    try:
        out = subprocess.run(["nvidia-smi", "--query-compute-apps=pid", "--format=csv,noheader"], capture_output=True,
                             text=True, timeout=30).stdout
        return [x.strip() for x in out.splitlines() if x.strip()]
    except Exception as exc:
        return [f"nvidia-smi failed: {type(exc).__name__}"]


def _sup_stop(pgid):
    """Stop the old server's group; True once it and every GPU compute process are gone."""
    for sig, wait_s in ((signal.SIGTERM, 45), (signal.SIGKILL, 180)):
        try:
            os.killpg(pgid, sig)
        except ProcessLookupError:
            pass
        except Exception as exc:
            _sup_log(f"killpg({pgid}, {sig.name}): {type(exc).__name__}: {exc}")
        t = time.time()
        while time.time() - t < wait_s:
            if not _sup_group_alive(pgid) and not _sup_gpu_apps():
                return True
            time.sleep(3)
    return False


def _sup_models_ok():
    try:
        with urllib.request.urlopen(f"{MANH_BASE_URL}/models", timeout=5) as u:
            return any(m.get("id") == "dsv41" for m in json.loads(u.read())["data"])
    except Exception:
        return False


def _sup_metrics():
    """(generation tokens total, running + waiting requests) from /metrics, or None if unreadable."""
    try:
        with urllib.request.urlopen(MANH_BASE_URL.removesuffix("/v1") + "/metrics", timeout=10) as u:
            text = u.read().decode("utf-8", "replace")
    except Exception:
        return None
    gen = busy = 0.0
    for line in text.splitlines():
        if line.startswith("vllm:generation_tokens_total"):
            gen += float(line.rsplit(" ", 1)[1])
        elif line.startswith(("vllm:num_requests_running", "vllm:num_requests_waiting")):
            busy += float(line.rsplit(" ", 1)[1])
    return gen, busy


def _sup_relaunch(n):
    root = f"{WORK}/dsv41_logs/r{n}"
    cmd = list(_cmd)
    cmd[cmd.index("--output-root") + 1] = root
    t = time.time()
    r = subprocess.run(cmd, cwd=WORK, env=_env, stdin=subprocess.DEVNULL, capture_output=True, text=True,
                       errors="replace", timeout=900)
    if r.returncode != 0:
        raise RuntimeError(f"launcher exited {r.returncode}: {(r.stdout + r.stderr)[-1500:]}")
    pid = int(json.load(open(f"{root}/server/launch.json"))["pid"])
    log = f"{root}/server/server.log"
    while time.time() - t < E066_UP_TIMEOUT_S:
        if _sup_models_ok():
            return pid, log, time.time() - t
        if not _manh_alive(pid):
            raise RuntimeError(f"server exited during startup; {log}")
        time.sleep(5)
    try:
        os.killpg(pid, signal.SIGKILL)
    except Exception:
        pass
    raise RuntimeError(f"server not up after {time.time() - t:.0f}s; {log}")


def _sup_loop():
    global MANH_SERVER_PID, MANH_SERVER_LOG
    start, prog_t, prog_gen = 0, time.time(), -1.0
    while not SUPERVISOR["stop"]:
        time.sleep(5)
        pid = MANH_SERVER_PID
        why = None
        if not pid or not _manh_alive(pid):
            why = f"server pid {pid} exited"
        elif E066_STALL_S > 0:
            m = _sup_metrics()
            if m is not None and (m[0] > prog_gen or m[1] == 0):  # progress, or idle (nothing to generate)
                prog_t, prog_gen = time.time(), m[0]
            elif time.time() - prog_t > E066_STALL_S:
                why = (f"no generated tokens for {time.time() - prog_t:.0f}s with requests running"
                       if m is not None else f"/metrics unreadable for {time.time() - prog_t:.0f}s")
        if why is None or SUPERVISOR["stop"]:
            continue
        left = T0 + float(os.environ.get("ARC_STOP_NEW_S", 39000)) - time.time()
        if SUPERVISOR["restarts"] >= E066_MAX_RESTARTS or left < E066_RESTART_MIN_LEFT_S:
            _sup_log(f"{why}; no restart ({SUPERVISOR['restarts']}/{E066_MAX_RESTARTS} used, {left / 60:.0f} min to "
                     f"stop-new)")
            _sup_write(start=start, state="failed", error=f"{why}; no restart left")
            return
        start += 1
        SUPERVISOR["restarts"] += 1
        _sup_write(start=start, state="restarting", error=why)
        _sup_log(f"{why}; restart {start}/{E066_MAX_RESTARTS} ({left / 60:.0f} min to stop-new)")
        if not _sup_stop(pid):
            _sup_log(f"old server group {pid} or GPU processes still there: {_sup_gpu_apps()}; giving up")
            _sup_write(start=start, state="failed", error="old server would not stop")
            return
        try:
            new_pid, log, up_s = _sup_relaunch(start)
        except Exception as exc:
            _sup_log(f"relaunch {start} failed: {type(exc).__name__}: {exc}")
            MANH_SERVER_PID = None
            continue  # counts as another death: retried while restarts and time are left
        MANH_SERVER_PID, MANH_SERVER_LOG = new_pid, log
        prog_t, prog_gen = time.time(), -1.0
        _sup_write(start=start, state="up", pid=new_pid, log=log, up_s=round(up_s))
        _sup_log(f"start {start} up after {up_s:.0f}s: pid {new_pid}, log {log}")


if MODE != "fallback" and globals().get("MANH_SERVER_PID"):
    os.makedirs(os.path.dirname(SUPERVISOR_STATUS), exist_ok=True)
    _sup_write(start=0, state="up", pid=MANH_SERVER_PID, log=MANH_SERVER_LOG, up_s=round(time.time() - _t))
    SUPERVISOR["thread"] = threading.Thread(target=_sup_loop, daemon=True)
    SUPERVISOR["thread"].start()
    _sup_log(f"watching pid {MANH_SERVER_PID} (max {E066_MAX_RESTARTS} restarts, stall {E066_STALL_S:.0f}s, "
             f"status {SUPERVISOR_STATUS})")
else:
    SUPERVISOR_STATUS = ""  # cell 3: no status file, the solver keeps S25 (no restarts)
```

```python
# Cell 2b (smoke, commit runs only): the tool sandbox at full load, before the solver starts. Starts as many sandbox
# sessions as the solver runs workers (each its own process with a 4 GiB address-space cap, 64 MB file cap, nice 10 and
# oom_score_adj 1000, like solver.session_factory), runs a typical ARC cell in every one, then checks that an oversized
# allocation fails inside its session, that a runaway cell is interrupted and the session restarts, and that closing
# every session returns host memory. Commit run: a failure RAISES ("SANDBOX CHECK FAILED"); the competition rerun skips
# this cell. Uses cell 0's KIT, MODE and RERUN, and cell 2's TOOL_PY.
import os
import sys
import threading
import time


def sandbox_meminfo_gib():
    out = {}
    with open("/proc/meminfo") as f:
        for line in f:
            key, value = line.split(":", 1)
            if key in ("MemAvailable", "AnonPages", "Shmem"):
                out[key] = int(value.split()[0]) / 2**20
    return out


def sandbox_pss_gib(pids):
    total = 0
    for pid in pids:
        try:
            with open(f"/proc/{pid}/smaps_rollup") as f:
                total += sum(int(l.split()[1]) for l in f if l.startswith("Pss:"))
        except OSError:
            pass
    return total / 2**20


def sandbox_check(kit, python, sessions, mem_mb=4096, timeout=30, alloc_gib=6):
    """Returns (report lines, failures)."""
    sys.path.insert(0, f"{kit}/repo/evals")
    from arc_harness.executor import LocalExecutor
    train = [{"input": [[0, 1, 0], [1, 1, 1]], "output": [[1, 0, 1], [0, 0, 0]]}]
    test = [[1, 0, 0], [0, 1, 1]]
    cell = ("import numpy as np\nfrom scipy import ndimage\ng = np.array(test_input)\n"
            "lab, n = ndimage.label(g)\nprint(n, (1 - g).tolist())")
    lines, failures, results, lock = [], [], [], threading.Lock()
    before = sandbox_meminfo_gib()
    exs = [LocalExecutor(train, test, timeout, 16000, memory_mb=mem_mb, python=python, style="ipython",
                         cpu_seconds=7200) for _ in range(sessions)]

    def run(ex):
        t0 = time.time()
        try:
            r = ex.execute(cell)
            ok = r.get("process_status") in (None, "ok", "completed") and "[[0, 1, 1]" in (r.get("stdout") or "")
            with lock:
                results.append((ok, time.time() - t0, (r.get("stderr") or "")[-200:]))
        except Exception as exc:
            with lock:
                results.append((False, time.time() - t0, f"{type(exc).__name__}: {exc}"))

    threads = [threading.Thread(target=run, args=(ex,)) for ex in exs]
    t0 = time.time()
    for th in threads:
        th.start()
    for th in threads:
        th.join(300)
    pids = [ex.proc.pid for ex in exs if getattr(ex, "proc", None)]
    loaded = sandbox_meminfo_gib()
    good = sum(1 for ok, _, _ in results if ok)
    slow = max((dt for _, dt, _ in results), default=0)
    lines.append(f"{good}/{sessions} sessions ran a numpy/scipy cell in {time.time() - t0:.0f} s (slowest {slow:.0f} s); "
                 f"sandbox PSS {sandbox_pss_gib(pids):.2f} GiB; host anon {before['AnonPages']:.1f} -> "
                 f"{loaded['AnonPages']:.1f} GiB, MemAvailable {before['MemAvailable']:.1f} -> {loaded['MemAvailable']:.1f} GiB")
    if good < sessions:
        failures.append(f"{sessions - good} sessions failed: {[e for ok, _, e in results if not ok][:2]}")

    # an allocation over the session's cap must fail inside the session, which stays usable
    ex = exs[0]
    r = ex.execute(f"x = bytearray({alloc_gib} * 2**30)\nprint('allocated')")
    text = (r.get("stdout") or "") + (r.get("stderr") or "")
    contained = "allocated" not in text and ("MemoryError" in text or r.get("state_reset"))
    after = ex.execute("print(6 * 7)")
    lines.append(f"{alloc_gib} GiB allocation in one session: {'blocked' if contained else 'NOT blocked'} "
                 f"({text.strip().splitlines()[-1][:90] if text.strip() else r.get('process_status')}); session after: "
                 f"{(after.get('stdout') or '').strip() or after.get('process_status')}")
    if not contained or "42" not in (after.get("stdout") or ""):
        failures.append("memory cap did not contain the allocation or the session did not recover")

    # a runaway cell is interrupted at the timeout and the session keeps working
    ex = exs[1]
    t1 = time.time()
    r = ex.execute("while True:\n    pass")
    after = ex.execute("print(6 * 7)")
    lines.append(f"runaway cell: {r.get('process_status')} after {time.time() - t1:.0f} s; session after: "
                 f"{(after.get('stdout') or '').strip() or after.get('process_status')}")
    if "42" not in (after.get("stdout") or ""):
        failures.append("session did not recover after a timeout")

    for ex in exs:
        ex.close()
    time.sleep(3)
    alive = [p for p in pids if os.path.exists(f"/proc/{p}")]
    end = sandbox_meminfo_gib()
    lines.append(f"closed: {len(alive)} sandbox processes left; host anon {end['AnonPages']:.1f} GiB, "
                 f"MemAvailable {end['MemAvailable']:.1f} GiB")
    if alive:
        failures.append(f"{len(alive)} sandbox processes outlived close()")
    if end["AnonPages"] > before["AnonPages"] + 1.0:
        failures.append(f"host anon memory not returned ({before['AnonPages']:.1f} -> {end['AnonPages']:.1f} GiB)")
    return lines, failures


if not RERUN and MODE != "fallback":
    try:
        _sb_lines, _sb_fail = sandbox_check(KIT, TOOL_PY, int(os.environ.get("ARC_SMOKE_SANDBOXES", "56")),
                                            int(os.environ.get("ARC_EXEC_MEM_MB", "4096")),
                                            int(os.environ.get("ARC_SMOKE_SANDBOX_TIMEOUT", "30")))
    except Exception as exc:
        _sb_lines, _sb_fail = [], [f"{type(exc).__name__}: {exc}"]
    for _l in _sb_lines:
        print("sandbox check:", _l, flush=True)
    if _sb_fail:
        print("=" * 100, f"SANDBOX CHECK FAILED: {_sb_fail}", "=" * 100, sep="\n", flush=True)
        raise RuntimeError(f"SANDBOX CHECK FAILED: {_sb_fail}")
    print("SANDBOX CHECK OK", flush=True)
```

```python
# Cell 2c (B32-M smoke, commit runs only): the API contract of the solver's server (his vLLM), checked in a
# background thread once it answers (E062 l4/api_check.py), so it costs no wall time. Required (the report cell
# fails the commit run otherwise): /v1/models serves dsv41 with max_model_len >= ARC_CONTEXT_LIMIT; /tokenize +
# /detokenize render a tool transcript; token-id /v1/completions returns token_ids (--token-exact); /metrics has
# vllm:kv_cache_usage_perc (KV gate). Recorded only: previous-turn thinking via `reasoning_content` (B32's harness
# key; upstream reads `reasoning`: only the rare full re-renders, ~0.1% of turns, lose it) and via `reasoning`, and
# a chat tool call (the harness parses tool calls itself on the token-exact path). Writes smoke_contract.json.
import re, sys, threading, urllib.request

CONTRACT_OUT = f"{WORK}/smoke_contract.json"


def contract_run(base, model="dsv41", wait_s=3600):
    res, fail = {}, []

    def post(path, body, timeout=600):
        req = urllib.request.Request(base + path, data=json.dumps(body).encode(),
                                     headers={"Content-Type": "application/json"})
        with urllib.request.urlopen(req, timeout=timeout) as r:
            return json.loads(r.read())
    t = time.time()
    models = None
    while time.time() - t < wait_s:
        try:
            with urllib.request.urlopen(base + "/v1/models", timeout=5) as r:
                models = json.loads(r.read())["data"]
            break
        except Exception:
            time.sleep(10)
    res["server_up_at_s"] = round(time.time() - T0)
    if not models:
        return {**res, "fail": [f"server never answered /v1/models within {wait_s}s"], "done": True}
    res["models"] = [(m.get("id"), m.get("max_model_len")) for m in models]
    ctx = int(os.environ.get("ARC_CONTEXT_LIMIT", "163840"))
    if not any(m.get("id") == model and (m.get("max_model_len") or 0) >= ctx for m in models):
        fail.append(f"/v1/models {res['models']}: need {model} with max_model_len >= {ctx}")
    kw = {"thinking": True, "reasoning_effort": 50}
    tools = [{"type": "function", "function": {
        "name": "run_python", "description": "Run Python code in a persistent session and return stdout.",
        "parameters": {"type": "object", "properties": {"code": {"type": "string"}}, "required": ["code"]}}}]
    mark = "ZEBRA-THINKING-MARKER-42"
    try:
        for field in ("reasoning_content", "reasoning"):
            msgs = [{"role": "user", "content": "What is 2+3? Use the tool."},
                    {"role": "assistant", "content": "", field: f"I should compute it. {mark}",
                     "tool_calls": [{"id": "c1", "type": "function", "function": {
                         "name": "run_python", "arguments": json.dumps({"code": "print(2+3)"})}}]},
                    {"role": "tool", "tool_call_id": "c1", "content": "5"}]
            tk = post("/tokenize", {"model": model, "messages": msgs, "tools": tools, "chat_template_kwargs": kw,
                                    "add_generation_prompt": True})
            dt = post("/detokenize", {"model": model, "tokens": tk["tokens"]})
            res[f"thinking_via_{field}"] = mark in dt["prompt"]
            res[f"tokenize_{field}_n"] = tk.get("count")
    except Exception as exc:
        fail.append(f"/tokenize or /detokenize: {type(exc).__name__}: {exc}"[:300])
    try:
        msgs = [{"role": "system", "content": "You solve tasks with the run_python tool."},
                {"role": "user", "content": "Compute the sum of the squares of 1..10 with run_python."}]
        tk = post("/tokenize", {"model": model, "messages": msgs, "tools": tools, "chat_template_kwargs": kw,
                                "add_generation_prompt": True})
        r = post("/v1/completions", {"model": model, "prompt": tk["tokens"], "max_tokens": 64, "temperature": 0,
                                     "return_token_ids": True, "skip_special_tokens": False,
                                     "cache_salt": "b32m-contract"})
        res["completion_token_ids"] = len(r["choices"][0].get("token_ids") or [])
        if not res["completion_token_ids"]:
            fail.append("token-id /v1/completions returned no token_ids")
    except Exception as exc:
        fail.append(f"token-id /v1/completions: {type(exc).__name__}: {exc}"[:300])
    try:
        with urllib.request.urlopen(base + "/metrics", timeout=10) as r:
            m = r.read().decode(errors="replace")
        res["metrics"] = {k: bool(re.search(rf"^vllm:{k}\b", m, re.M)) for k in (
            "kv_cache_usage_perc", "num_requests_running", "num_requests_waiting", "generation_tokens_total",
            "prefix_cache_hits_total", "num_preemptions_total")}
        if not res["metrics"]["kv_cache_usage_perc"]:
            fail.append("/metrics lacks vllm:kv_cache_usage_perc (KV gate)")
    except Exception as exc:
        fail.append(f"/metrics: {type(exc).__name__}: {exc}"[:300])
    try:
        c = post("/v1/chat/completions", {"model": model, "messages": msgs, "tools": tools, "temperature": 0,
                                          "max_tokens": 4000, "chat_template_kwargs": kw})
        res["chat_tool_calls"] = len(c["choices"][0]["message"].get("tool_calls") or [])
    except Exception as exc:
        res["chat_tool_calls"] = f"{type(exc).__name__}"
    if os.environ.get("ARC_GRID_FORMAT") == "text+image":  # E066: the image harness runs on the chat path
        if not (isinstance(res["chat_tool_calls"], int) and res["chat_tool_calls"] >= 1):
            fail.append(f"chat tool calls not parsed ({res['chat_tool_calls']}): the image harness needs the "
                        "server's --tool-call-parser deepseek_v41 --enable-auto-tool-choice")
        try:  # one token-aligned grid image next to its text, as the harness sends it
            sys.path.insert(0, f"{KIT}/repo/evals")
            from arc_harness.grid_image import aligned_cell, data_uri
            g = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
            text = {"type": "text", "text": "Describe this 3x3 grid in one sentence."}
            img = {"type": "image_url", "image_url": {"url": data_uri(g, aligned_cell(g, "deepseek"))}}
            n = []
            for content in ([text], [text, img]):
                c = post("/v1/chat/completions", {"model": model, "messages": [{"role": "user", "content": content}],
                                                  "temperature": 0, "max_tokens": 16, "chat_template_kwargs": kw})
                n.append(c["usage"]["prompt_tokens"])
            res["image_prompt_tokens"] = n[1] - n[0]
            if res["image_prompt_tokens"] <= 0:
                fail.append(f"an image added no prompt tokens ({n}): the server dropped it")
        except Exception as exc:
            fail.append(f"image chat request: {type(exc).__name__}: {exc}"[:300])
    return {**res, "fail": fail, "done": True}


def contract_thread():
    try:
        out = contract_run(f"http://127.0.0.1:{os.environ.get('ARC_PORT', '8100')}")
    except Exception as exc:
        out = {"fail": [f"contract check crashed: {type(exc).__name__}: {exc}"], "done": True}
    with open(CONTRACT_OUT, "w") as f:
        json.dump(out, f, indent=1)
    print("B32-M contract:", out, flush=True)


if not RERUN and MODE != "fallback":
    _ct_thread = threading.Thread(target=contract_thread, daemon=True)  # E066: cell 3w waits for it
    _ct_thread.start()
    print("B32-M contract check started (background)", flush=True)
```

```python
# Cell 3: the solver (tool harness, deadlines, watchdog, incremental submission). B32-M + Manh's server: it calls
# his server (cell 2m) at ARC_EXTERNAL_SERVER = MANH_BASE_URL (S23: no launch, no stop, no restart; its log gets
# "server up after ...s (external)" with the KV line of ARC_EXTERNAL_SERVER_LOG). The model dir is his server's
# (the tokenizer for the loop guard); cell 2m already waited for the staging marker.
if MODE != "fallback" and globals().get("MANH_SERVER_PID"):
  try:
    import signal, threading
    env = dict(BASE_ENV)
    env.update(ARC_KIT=KIT, ARC_MODEL_DIR=MANH_MODEL, ARC_STAGE_DONE="", ARC_EXTERNAL_SERVER=MANH_BASE_URL,
               ARC_EXTERNAL_SERVER_LOG=MANH_SERVER_LOG,  # E066: + S26, follow cell 2s through restarts
               ARC_MAX_SERVER_RESTARTS="3" if globals().get("SUPERVISOR_STATUS") else "0",
               ARC_EXTERNAL_SERVER_STATUS=globals().get("SUPERVISOR_STATUS") or "", ARC_TOOL_PYTHON=TOOL_PY,
               ARC_SCRATCH=SCRATCH)
    if MODE == "smoke":
        env.setdefault("ARC_TASK_LIMIT", "8")
        for k, v in (("ARC_STOP_NEW_S", "4800"), ("ARC_FORCE_AT_S", "5400"), ("ARC_HARD_STOP_S", "5700")):
            env.setdefault(k, v)
    p = subprocess.Popen([PY, "-u", f"{KIT}/solver_dsv41.py"], env=env, stdout=subprocess.PIPE,
                         stderr=subprocess.STDOUT, text=True, errors="replace", start_new_session=True)
    # Independent kill switch: whatever hangs inside the solver, its whole session dies at T0 + ARC_KILL_AT_S
    # (default 11h45; the solver's own hard stop is 11h30), so cell 4 and the notebook end before Kaggle's 12 h.
    def _kill():
        try:
            os.killpg(p.pid, signal.SIGKILL)
            print("kill switch: solver session killed", flush=True)
        except Exception:
            pass
    _timer = threading.Timer(max(0.0, T0 + float(os.environ.get("ARC_KILL_AT_S", 42300)) - time.time()), _kill)
    _timer.daemon = True
    _timer.start()
    for line in p.stdout:
        print(line, end="", flush=True)
    p.wait()
    _timer.cancel()
    print("solver exit code", p.returncode, flush=True)
  except Exception as exc:
    print("solver cell failed:", type(exc).__name__, exc, flush=True)
```
